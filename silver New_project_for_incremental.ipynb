```python
bronze_df = spark.read.table("bronze_customer_data")

display(bronze_df)
```

```python
from pyspark.sql import functions as f

silvers_df = bronze_df \
    .withColumn("customer_id", f.trim(f.col("customer_id"))) \
    .withColumn("customer_name", f.trim(f.col("customer_name"))) \
    .withColumn("city", f.upper(f.trim(f.col("city")))) \
    .withColumn("account_status", f.upper(f.trim(f.col("account_status"))))

display(silvers_df)
```

```python
silvers_df.write \
    .format("delta") \
    .mode("append") \
    .saveAsTable("silver_customer_data")
```
