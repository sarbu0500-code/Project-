```python
silver_df = spark.read.table("silver_customer_data")

display(silver_df)
```

```python
from pyspark.sql import functions as f
gold_df = silver_df.groupBy("city").agg(
    f.count("customer_id").alias("total_customers"),
    f.sum("account_balance").alias("total_balance"),
    f.sum(
        f.when(f.col("account_status") == "ACTIVE", 1).otherwise(0)
    ).alias("active_customers")
)

display(gold_df)
```

```python
gold_df.write \
    .format("delta") \
    .mode("append") \
    .saveAsTable("gold_customer_summary")
```
